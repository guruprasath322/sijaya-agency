## Shallow Parsing

```python
import nltk
from nltk import RegexpParser
text="The Avengers movie was an absolute masterpiece from start to finish."
text_tokens =nltk.word_tokenize(text)
tagged_sent=nltk.pos_tag(text_tokens)
chunk_grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>}
"""
rc=RegexpParser(chunk_grammar)
parsed_sent = rc.parse(tagged_sent)
print(parsed_sent)
parsed_sent.pretty_print()
```

## Constituency Parsers

```python
import nltk
from nltk.chunk import RegexpParser
test_sentence = "The Avengers movie was an absolute masterpiece from start to finish."

tokens = nltk.word_tokenize(test_sentence)
tagged_sentence = nltk.pos_tag(tokens)

print("\nPOS Tags:")
print(tagged_sentence)

grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>}
    PP: {<IN><NP>}
    VP: {<VB.*><NP|PP>*}
    ADJP: {<JJ>}
"""

parser = RegexpParser(grammar)
tree = parser.parse(tagged_sentence)
print("\nCONSTITUENCY PARSING")
print("\nParse Tree:")
print(tree)
print("\nPretty Print:")
tree.pretty_print()
```
